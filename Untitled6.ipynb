***QUESTION NUMBER 1***

What is the age distribution?




```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Load your dataset:
# Replace the line below with: df = pd.read_csv("exact_filename.csv")
df = pd.DataFrame({"age": np.random.randint(40, 95, size=299)})

# 2. Convert age column to a NumPy array for calculations
age_arr = df["age"].dropna().to_numpy()

print("Age Count:", age_arr.size)
print("Age Min:", np.min(age_arr))
print("Age Max:", np.max(age_arr))
print("Age Mean:", round(float(np.mean(age_arr)), 2))
print("Age Median:", float(np.median(age_arr)))
print("Age Std Dev:", round(float(np.std(age_arr, ddof=1)), 2))

# 3. Age distribution across 10-year bins using Pandas
age_bins = pd.cut(df["age"], bins=[35, 45, 55, 65, 75, 85, 100])
print("\nAge Distribution by Bracket (sorted):")
print(age_bins.value_counts().sort_index())

# 4. Chart 1: Age Histogram
plt.figure(figsize=(8, 5))
plt.hist(age_arr, bins=12, edgecolor="black")
plt.title(f"Age Distribution of Patients (n = {age_arr.size})")
plt.xlabel("Age (years)")
plt.ylabel("Number of Records")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

***Question Number2***       
       Count records by death_event





```python
import re
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Clean column names (strip spaces & lowercase) without altering original df
col_map = {
    re.sub(r"[\s_]+", "_", col.strip().lower()): col for col in df.columns
}

target_key = "death_event"
if target_key not in col_map:
    raise KeyError(
        f"Column 'death_event' not found. Available columns: {df.columns.tolist()}"
    )

col_name = col_map[target_key]
s = df[col_name].dropna()

# Count records by death_event (0 = survived / no event, 1 = event occurred)
death_counts = s.value_counts().sort_index()
death_pct = (s.value_counts(normalize=True).sort_index() * 100).round(2)

summary_death = pd.DataFrame({
    "record_count": death_counts,
    "percentage": death_pct,
})
print(f"Records by {col_name}:")
print(summary_death)

# Chart 2: death_event Bar Chart
labels = ["0 (No Death Event)", "1 (Death Event)"]
c0 = int(death_counts.get(0, 0))
c1 = int(death_counts.get(1, 0))
p0 = float(death_pct.get(0, 0.0))
p1 = float(death_pct.get(1, 0.0))

counts = np.array([c0, c1])
pcts = np.array([p0, p1])

upper_y = max(float(counts.max()) * 1.15, 1.0)
offset = upper_y * 0.015

plt.figure(figsize=(7, 5))
plt.bar(labels, counts, edgecolor="black", zorder=3)
plt.title(f"Record Count by {col_name}")
plt.xlabel(col_name)
plt.ylabel("Number of Records")

for i, (v, pct) in enumerate(zip(counts, pcts)):
    plt.text(
        i,
        v + offset,
        f"{v} ({pct:.1f}%)",
        ha="center",
        va="bottom",
        fontweight="bold",
        zorder=4,
    )

plt.ylim(0, upper_y)
plt.grid(axis="y", linestyle="--", alpha=0.5, zorder=0)
plt.tight_layout()
plt.show()
```

Question number 4

 Compare average ejection_fraction and serum_creatinine across death_event groups.



```python
import re
import pandas as pd


def resolve_columns(df: pd.DataFrame, required_keys: list[str]) -> dict[str, str]:
    col_map = {
        re.sub(r"[\s_]+", "_", str(col).strip().lower()): col
        for col in df.columns
    }
    resolved = {}
    missing = []
    for key in required_keys:
        norm_key = re.sub(r"[\s_]+", "_", key.strip().lower())
        if norm_key in col_map:
            resolved[key] = col_map[norm_key]
        else:
            missing.append(key)

    if missing:
        raise KeyError(
            f"Missing column(s) {missing}. Available columns: {df.columns.tolist()}"
        )
    return resolved


cols = resolve_columns(
    df, ["death_event", "ejection_fraction", "serum_creatinine", "age"]
)

death_col = cols["death_event"]
ef_col = cols["ejection_fraction"]
sc_col = cols["serum_creatinine"]
age_col = cols["age"]

# Group by death_event and compute mean (and count) for ejection_fraction and serum_creatinine
group_comparison = (
    df.groupby(death_col, dropna=True)[[ef_col, sc_col]]
    .agg(["count", "mean"])
    .sort_index()
)

print(f"Average {ef_col} and {sc_col} across {death_col} groups:")
print(group_comparison.round(3))

# Top 5 highest serum_creatinine among death_event == 1
high_creatinine_deaths = (
    df[df[death_col] == 1][[age_col, ef_col, sc_col, death_col]]
    .sort_values(by=sc_col, ascending=False)
    .head(5)
)
print(f"\nTop 5 highest {sc_col} records where {death_col} == 1:")
print(high_creatinine_deaths)
```

Question number 5                                     
Count records with and without diabetes, high blood pressure and anaemia.

```python
binary_cols = ["diabetes", "high_blood_pressure", "anaemia"]

for col in binary_cols:
    counts = df[col].value_counts().sort_index()
    without_count = int(counts.get(0, 0))
    with_count = int(counts.get(1, 0))
    print(f"{col}: Without (0) = {without_count} | With (1) = {with_count}")

# Combined summary table using Pandas value_counts
condition_summary = pd.DataFrame({
    col: df[col].value_counts().sort_index() for col in binary_cols
}).T.rename(columns={0: "without_condition (0)", 1: "with_condition (1)"})

print("\nSummary Table of Conditions:")
print(condition_summary)
```

```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def load_heart_failure_df(filename="heart_failure_clinical_records_dataset.csv"):
    # 1. Exact matches in current dir, /content, or Drive
    exact_candidates = [
        filename,
        os.path.join("/content", filename),
        os.path.join("/content/drive/MyDrive", filename),
    ]
    for path in exact_candidates:
        if os.path.exists(path):
            print(f"Loaded dataset from: {path}")
            return pd.read_csv(path)

    # 2. Check if any CSV file exists in current folder or /content (e.g., renamed or "(1).csv")
    csv_candidates = glob.glob("*.csv") + glob.glob("/content/*.csv")
    for path in csv_candidates:
        if os.path.isfile(path):
            print(f"Loaded found CSV file: {path}")
            return pd.read_csv(path)

    # 3. Non-blocking fallback dataset so the cell never hangs
    print("No CSV file found in directory. Using a fallback sample dataset so EDA runs immediately.")
    rng = np.random.default_rng(42)
    n = 299
    age = np.clip(np.round(rng.normal(60.8, 11.9, size=n)), 40, 95).astype(int)
    sex = rng.integers(0, 2, size=n)
    anaemia = rng.integers(0, 2, size=n)
    diabetes = rng.integers(0, 2, size=n)
    high_blood_pressure = rng.integers(0, 2, size=n)
    smoking = rng.integers(0, 2, size=n)
    creatinine_phosphokinase = rng.integers(23, 2500, size=n)
    ejection_fraction = np.clip(np.round(rng.normal(38.1, 11.8, size=n), -1), 14, 80).astype(int)
    platelets = np.round(np.clip(rng.normal(263358, 97804, size=n), 25100, 850000), 1)
    serum_creatinine = np.round(np.clip(rng.lognormal(mean=0.25, sigma=0.55, size=n), 0.5, 9.4), 2)
    serum_sodium = np.clip(rng.integers(113, 149, size=n), 113, 148)
    time = rng.integers(4, 286, size=n)
    death_event = rng.binomial(1, 0.32, size=n)

    return pd.DataFrame(
        {
            "age": age,
            "anaemia": anaemia,
            "creatinine_phosphokinase": creatinine_phosphokinase,
            "diabetes": diabetes,
            "ejection_fraction": ejection_fraction,
            "high_blood_pressure": high_blood_pressure,
            "platelets": platelets,
            "serum_creatinine": serum_creatinine,
            "serum_sodium": serum_sodium,
            "sex": sex,
            "smoking": smoking,
            "time": time,
            "death_event": death_event,
        }
    )


# 1. Load dataset & normalize column names to lowercase
df = load_heart_failure_df("heart_failure_clinical_records_dataset.csv")
df.columns = df.columns.str.strip().str.lower()

# 2. Basic structure and summary statistics
print("--- HEAD ---")
print(df.head())

print("\n--- TAIL ---")
print(df.tail())

print("\n--- SHAPE ---")
print(df.shape)

print("\n--- INFO ---")
print(df.info())

print("\n--- DESCRIBE ---")
print(df.describe())

# 3. Check missing values and duplicate rows
print("\n--- MISSING VALUES PER COLUMN ---")
print(df.isnull().sum())
print("Total missing values:", int(df.isnull().sum().sum()))

print("\n--- DUPLICATE ROWS ---")
print("Number of duplicate rows:", int(df.duplicated().sum()))

# 4. NumPy array calculations on 'age'
age_arr = df["age"].to_numpy(dtype=float)
age_mean = np.mean(age_arr)
age_median = np.median(age_arr)
age_std = np.std(age_arr, ddof=1)
age_q25, age_q75 = np.percentile(age_arr, [25, 75])
age_iqr = age_q75 - age_q25

print("\n--- NUMPY AGE CALCULATIONS ---")
print(f"Mean Age: {age_mean:.2f}")
print(f"Median Age: {age_median:.2f}")
print(f"Sample Std Dev: {age_std:.2f}")
print(f"IQR (Q3 - Q1): {age_iqr:.2f} (Q1={age_q25:.1f}, Q3={age_q75:.1f})")

# 5. Pandas value_counts(), filtering, sorting, and groupby()
print("\n--- DEATH_EVENT VALUE COUNTS ---")
print(df["death_event"].value_counts(dropna=False))
print("\n--- DEATH_EVENT PROPORTIONS ---")
print(df["death_event"].value_counts(normalize=True).round(4))

older_patients_sorted = df[df["age"] >= 70].sort_values(
    by=["age", "serum_creatinine"], ascending=[False, False]
)
print("\n--- FILTERED (AGE >= 70) & SORTED TOP 5 ---")
cols_to_show = [c for c in ["age", "sex", "serum_creatinine", "death_event"] if c in df.columns]
print(older_patients_sorted[cols_to_show].head())

print("\n--- GROUPBY DEATH_EVENT SUMMARY ---")
agg_kwargs = {
    "patient_count": ("age", "count"),
    "mean_age": ("age", "mean"),
    "median_age": ("age", "median"),
}
if "ejection_fraction" in df.columns:
    agg_kwargs["mean_ejection_fraction"] = ("ejection_fraction", "mean")
if "serum_creatinine" in df.columns:
    agg_kwargs["mean_serum_creatinine"] = ("serum_creatinine", "mean")

grouped_summary = df.groupby("death_event", as_index=False).agg(**agg_kwargs)
print(grouped_summary.round(2))

# 6. Charts: (a) death_event bar chart & (b) age histogram
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

death_counts = df["death_event"].value_counts().sort_index()
axes[0].bar(
    death_counts.index.astype(str),
    death_counts.values,
    color=["#4C72B0", "#DD8452"],
    edgecolor="black",
)
axes[0].set_title("Death Event Distribution")
axes[0].set_xlabel("death_event (0 = No, 1 = Yes)")
axes[0].set_ylabel("Number of Records")
for idx, val in enumerate(death_counts.values):
    axes[0].text(idx, val + max(death_counts.values) * 0.015, str(val), ha="center", va="bottom", fontsize=10)

axes[1].hist(age_arr, bins=15, color="#55A868", edgecolor="black", alpha=0.85)
axes[1].axvline(age_mean, color="red", linestyle="--", linewidth=1.5, label=f"Mean ({age_mean:.1f})")
axes[1].axvline(age_median, color="navy", linestyle=":", linewidth=1.5, label=f"Median ({age_median:.1f})")
axes[1].set_title("Age Distribution")
axes[1].set_xlabel("Age (years)")
axes[1].set_ylabel("Frequency")
axes[1].legend()

plt.tight_layout()
plt.show()
```
